## Semantic Search Agent
Call the ReAct agent that uses the semantic search tool to find and recommend movies.

```python
from movie_agent.agent.semantic_search_agent import ask, ask_stream
```

```python
# Non-streaming call
result = ask("I want a movie about time travel and love")
print(result["answer"])
print(f"\nSession ID: {result['session_id']}")
```

```python
# Streaming call - iterate over the generator to see each step
print("=" * 60)
print("STREAMING: agent steps + answer")
print("=" * 60)

full_answer = ""
session_id = None

for event in ask_stream("a heist movie with a clever twist"):
    session_id = event["session_id"]

    if event["type"] == "tool_call":
        print(f"\n🛠️  TOOL CALL: {event['tool']}")
        print(f"   Input: {event['input']}")

    elif event["type"] == "tool_result":
        print(f"\n📋 TOOL RESULT ({event['tool']}):")
        print(f"   {event['output'][:300]}...")

    elif event["type"] == "token":
        full_answer += event["content"]
        print(event["content"], end="", flush=True)

print(f"\n\n{'=' * 60}")
print(f"Session ID: {session_id}")
```

```python
# Follow-up using the same session (agent remembers context)
print("Follow-up question (same session):\n")

for event in ask_stream("Tell me more about the second one", session_id=session_id):
    if event["type"] == "tool_call":
        print(f"\n🛠️  TOOL CALL: {event['tool']} -> {event['input']}")
    elif event["type"] == "tool_result":
        print(f"📋 RESULT: {event['output'][:200]}...\n")
    elif event["type"] == "token":
        print(event["content"], end="", flush=True)

print()
```
