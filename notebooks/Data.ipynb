```python
def explode_json_column(col, keys):
    exploded_table = df[[col]].set_index(df.id)[col].apply(lambda x: eval(x)).explode().apply(pd.Series)[keys]

    lookup_table = exploded_table.drop_duplicates().dropna()

    return exploded_table, lookup_table.reset_index(drop=True)
genres_exploded_table, genres_lookup_table = explode_json_column('genres', ['id', 'name'])
language_exploded_table, language_lookup_table = explode_json_column('spoken_languages', ['iso_639_1', 'name'])
companies_exploded_table, companies_lookup_table = explode_json_column('production_companies', ['id', 'name'])
countries_exploded_table, countries_lookup_table = explode_json_column('production_countries', ['iso_3166_1', 'name'])
```

```python
import psycopg2
from psycopg2.extras import execute_values
import pandas as pd

# Connect to PostgreSQL
conn = psycopg2.connect(
    host="localhost",
    port=5432,
    dbname="movie_agent",
    user="movie_agent",
    password="movie_agent_pass"
)
cur = conn.cursor()
print("Connected successfully!")
```

```python
# Enable pgvector extension
cur.execute("CREATE EXTENSION IF NOT EXISTS vector;")
conn.commit()

# Create the table
cur.execute("""
    DROP TABLE IF EXISTS movies;
    CREATE TABLE movies (
        id              INTEGER PRIMARY KEY,
        budget          BIGINT,
        homepage        TEXT,
        original_language TEXT,
        original_title  TEXT,
        overview        TEXT,
        popularity      REAL,
        release_date    DATE,
        revenue         BIGINT,
        runtime         REAL,
        status          TEXT,
        tagline         TEXT,
        title           TEXT NOT NULL,
        vote_average    REAL,
        vote_count      INTEGER,
        embeddings vector(384)
    );
""")
conn.commit()
print("Table 'movies' created successfully!")
```

```python
# Load and select columns
movies = pd.read_csv("./../data/tmdb_5000_movies.csv", encoding='utf-8')
movies['release_date'] = pd.to_datetime(movies['release_date']).dt.date

movies_selected = movies[['budget', 'homepage', 'id', 'original_language',
                          'original_title', 'overview', 'popularity',
                          'release_date', 'revenue', 'runtime', 'status',
                          'tagline', 'title', 'vote_average', 'vote_count']].copy()

# Replace NaN with None for proper NULL handling
movies_selected = movies_selected.where(pd.notna(movies_selected), None)
print(f"Rows to insert: {len(movies_selected)}")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
print("Model loaded.")
```

```python
# Generate embeddings for all overviews
overviews = movies_selected['overview'].fillna("").tolist()
embeddings = model.encode(overviews, show_progress_bar=True, batch_size=64)
print(f"Generated {len(embeddings)} embeddings")
```

```python
len(embeddings[0])
```

```python
# Prepare rows and insert
rows = [
    (
        int(row['id']),
        int(row['budget']) if row['budget'] is not None else None,
        row['homepage'],
        row['original_language'],
        row['original_title'],
        row['overview'],
        float(row['popularity']) if row['popularity'] is not None else None,
        row['release_date'],
        int(row['revenue']) if row['revenue'] is not None else None,
        float(row['runtime']) if row['runtime'] is not None else None,
        row['status'],
        row['tagline'],
        row['title'],
        float(row['vote_average']) if row['vote_average'] is not None else None,
        int(row['vote_count']) if row['vote_count'] is not None else None,
        embeddings[i].tolist()
    )
    for i, row in movies_selected.iterrows()
]

execute_values(cur, """
    INSERT INTO movies (id, budget, homepage, original_language, original_title,
                               overview, popularity, release_date, revenue, runtime,
                               status, tagline, title, vote_average, vote_count, embedding)
    VALUES %s
    ON CONFLICT (id) DO NOTHING
""", rows)

conn.commit()
print(f"Inserted {len(rows)} rows into movies.")
```
